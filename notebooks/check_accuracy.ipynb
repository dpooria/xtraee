Amplitude Matching score
$$
        e = \frac{\sum_i{|a_i^2 - b_i^2|}}{ N}
$$
$$
        e = \frac{\sum_i{a_i^2|a_i^2 - b_i^2|}}{\sum_i{a_i^2}}
$$
$$
        e = {\sum_i{\tilde{a}_i^2|\tilde{a}_i^2 - \tilde{b}_i^2|}}
$$
$$
ams = 1 -e
$$
